```python
# Square Roots

import math

def square_roots(numbers: list):  
  return [math.sqrt(num) for num in numbers]

```

```python
# Rows of stars

def rows_of_stars(numbers: list):
  
  return [('*' * num) for num in numbers]
```

```python
# Lengths

def lengths(lists: list):
  
  return [len(li) for li in lists]
```

```python
# Remove smaller than

def remove_smaller_than(numbers: list, limit: int):
  
  return [num for num in numbers if num >= limit]
```

```python
# Begin with vowel

def begin_with_vowel(words: list):
  
  return [word for word in words if word[0].lower() in 'aeiou']
```

```python
# Lottery Numbers

class LotteryNumbers:
  
  def __init__(self, lottery_week: int, correct_lottery_nums: list):
    self.lottery_week = lottery_week
    self.correct_lottery_nums = correct_lottery_nums
  
  def number_of_hits(self, test_numbers: list):
    return len([num for num in test_numbers if num in self.correct_lottery_nums])
    
  def hits_in_place(self, numbers: list):
    return [num if num in self.correct_lottery_nums else -1 for num in numbers]
  

  
```

```python
# Shopping List

class ShoppingList:
    def __init__(self):
        self.products = []

    def number_of_items(self):
        return len(self.products)

    def add(self, product: str, number: int):
        self.products.append((product, number))

    def __iter__(self):
        self.n = 0
        return self

    def __next__(self):
        if self.n < len(self.products):
            product = self.products[self.n]
            self.n += 1
            return product
        else:
            raise StopIteration\

def products_in_shopping_list(shopping_list, amount: int):
    return [item for item, quantity in shopping_list if quantity >= amount]

    
```

```python
# Cheaper Properties

class RealProperty:
    def __init__(self, rooms: int , square_meters: int, price_per_sqm: int, description: str):
        self.rooms = rooms
        self.square_meters = square_meters
        self.price_per_sqm = price_per_sqm
        self.description = description

    def bigger(self, compared_to):
        return self.square_meters > compared_to.square_meters

    def price_difference(self, compared_to):
        difference = abs((self.price_per_sqm * self.square_meters) - (compared_to.price_per_sqm * compared_to.square_meters))
        return difference

    def more_expensive(self, compared_to):
        difference = (self.price_per_sqm * self.square_meters) - (compared_to.price_per_sqm * compared_to.square_meters)
        return difference > 0


    def __repr__(self):
        return (f'RealProperty(rooms = {self.rooms}, square_meters = {self.square_meters}, ' + 
            f'price_per_sqm = {self.price_per_sqm}, description = {self.description})')

def cheaper_properties(properties: list, reference: RealProperty):
    return [(prop, prop.price_difference(reference)) for prop in properties if not prop.more_expensive(reference) and prop != reference]
```

```python
# Most Commons Words
def most_common_words(filename: str, lower_limit: int):
    my_dict = {}
    grammar = '!.,:;()-'
    
    with open(filename, 'r') as f:
        for row in f:
            for word in row.split():
                clean_word = word.strip(grammar)                
                if clean_word:
                    if clean_word not in my_dict:
                        my_dict[clean_word] = 1
                    else:
                        my_dict[clean_word] += 1
      
    final_dict = {word: num for (word, num) in my_dict.items() if num >= lower_limit}
    
    return final_dict
```

```python
# Recursive Sum
def recursive_sum(number: int):
  res = 0
  if number <= 1:
    return number
  else:
    while number >= 1:
      res += number
      number -= 1
      recursive_sum(number)
      
  return res

```

```python
# Balanced Brackets

def balanced_brackets(my_string: str):
    s = "".join([c for c in my_string if c in "()[]"])
    
    if s == "":
        return True
    
    if "()" in s:
        return balanced_brackets(s.replace("()", "", 1))
    elif "[]" in s:
        return balanced_brackets(s.replace("[]", "", 1))
    
    return False
```

```python
# Greatest Node

class Node:
    """ Class is modeling single node in binary tree """
    def __init__(self, value, left_child:'Node' = None, right_child:'Node' = None):
        self.value = value
        self.left_child = left_child
        self.right_child = right_child

def greatest_node(root: Node):

    max_val = root.value

    if root.left_child:
        left_max = greatest_node(root.left_child)
        if left_max > max_val:
            max_val = left_max
            
    if root.right_child:
        right_max = greatest_node(root.right_child)
        if right_max > max_val:
            max_val = right_max

    return max_val


            
```

```python
# Bosses and subordinates 

class Employee:
    def __init__(self, name: str):
        self.name = name
        self.subordinates = []

    def add_subordinate(self, employee: 'Employee'):
        self.subordinates.append(employee)


def count_subordinates(employee: Employee):
    
    count = 0
    count += len(employee.subordinates)
    
    for emp in employee.subordinates:
        if emp is not None:
            count += count_subordinates(emp)
    
    return count
```

```python
# Order_Book
class Task:
    identification = 0
    
    def __init__(self, description: str, programmer: str, workload: int):
        Task.identification += 1
        self.description = description
        self.programmer = programmer
        self.workload = workload
        self.status = 'NOT FINISHED'
        self.id = Task.identification 
    
    def is_finished(self):
        return self.status == 'FINISHED'
    
    def mark_finished(self):
        self.status = 'FINISHED'

    def __str__(self):
        return f'{self.id}: {self.description} ({self.workload} hours), programmer {self.programmer} {self.status}'
  
  
class OrderBook:
  
  def __init__(self):
    self.orders = []
  
  def add_order(self, description: str, programmer: str, workload: int):
    new_task = Task(description, programmer, workload)
    self.orders.append(new_task)
  
  def all_orders(self):
    return self.orders
  
  def programmers(self):
    names = []
    for order in self.orders:
      names.append(order.programmer)
    
    names.sort()
    
    return list(set(names))
    
  def mark_finished(self, id: int):
    found = False    
    try:
        for order in self.orders:
            if order.id == id:
                order.mark_finished()
                found = True
        
        if not found:
            raise ValueError(f"No task with ID {id} found.")            
    except ValueError:
        raise ValueError
    
  def finished_orders(self):
    return [order for order in self.orders if order.status == 'FINISHED']

  def unfinished_orders(self):
    return [order for order in self.orders if order.status != 'FINISHED']
  
  def status_of_programmer(self, programmer: str):
    complete = 0
    uncomplete = 0
    complete_hours = 0 
    uncomplete_hours = 0

    if programmer not in self.programmers():
        raise ValueError(f"No programmer named {programmer}")

    try:
        for order in self.orders:
            if order.programmer == programmer:
                if order.status == 'NOT FINISHED':
                    uncomplete += 1
                    uncomplete_hours += order.workload  
                elif order.status == 'FINISHED':
                    complete += 1
                    complete_hours += order.workload
    except ValueError:
        raise ValueError
    
    return (complete, uncomplete, complete_hours, uncomplete_hours)
        
```
